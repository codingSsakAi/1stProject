```python
import os

# 문제가 되는 환경 변수 제거
if 'MPLBACKEND' in os.environ:
    print("기존 MPLBACKEND:", os.environ['MPLBACKEND'])
    del os.environ['MPLBACKEND']
    print("MPLBACKEND 환경 변수 제거됨")
else:
    print("MPLBACKEND 환경 변수 없음")

# 이제 안전하게 import
import matplotlib
print("matplotlib import 성공!")
print("기본 백엔드:", matplotlib.get_backend())

# 적절한 백엔드로 설정
matplotlib.use('Agg')
print("새 백엔드:", matplotlib.get_backend())
```

```python
import sys
import subprocess

print("matplotlib 완전 제거 후 재설치...")
result = subprocess.run([sys.executable, '-m', 'pip', 'uninstall', 'matplotlib', '-y'], 
                       capture_output=True, text=True)
print("제거 결과:", result.stdout)

result = subprocess.run([sys.executable, '-m', 'pip', 'install', 'matplotlib'], 
                       capture_output=True, text=True)
print("재설치 결과:", result.stdout)
```

```python
# 필수 패키지들을 모두 함께 설치
packages = ['matplotlib', 'matplotlib-inline', 'ipympl', 'ipywidgets']
for package in packages:
    result = subprocess.run([sys.executable, '-m', 'pip', 'install', package], 
                           capture_output=True, text=True)
    print(f"{package} 설치:", "성공" if result.returncode == 0 else "실패")
```

```python
import os
# 환경 변수 정리
if 'MPLBACKEND' in os.environ:
    del os.environ['MPLBACKEND']

# 단계별 import 테스트
print("1. matplotlib import 시도...")
import matplotlib
print("   성공!")

print("2. pyplot import 시도...")
import matplotlib.pyplot as plt
print("   성공!")

print("3. 간단한 플롯 테스트...")
plt.figure()
plt.plot([1, 2, 3], [1, 4, 2])
plt.title("테스트 성공!")
plt.show()
```
